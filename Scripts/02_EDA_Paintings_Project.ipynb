Distribution of Artistic Movements

There is a minor imbalance in the data collection. We will use an imbalance-handling approach during training to resolve this.

Image Dimentions and Aspect Ratios

Since you will use ResNet50, it is important to understand the original image sizes before resizing them.

This cell extracts the width, height, and aspect ratio of every painting.

```python
import os
from PIL import Image

widths = []
heights = []
valid_images = []

for path in df["image_path"]:
    full_path = os.path.join("wikiart_images", path)

    try:
        with Image.open(full_path) as img:
            width, height = img.size

        widths.append(width)
        heights.append(height)
        valid_images.append(True)

    except (OSError, FileNotFoundError):
        widths.append(float("nan"))
        heights.append(float("nan"))
        valid_images.append(False)

df["width"] = widths
df["height"] = heights
df["aspect_ratio"] = df["width"] / df["height"]
df["valid_image"] = valid_images

print("Valid Images:", df["valid_image"].sum())
print("Missing/Unreadable Images:", (~df["valid_image"]).sum())

display(df[["width", "height", "aspect_ratio"]].describe())
```

Image Dimention Distributions

```python

fig, axes = plt.subplots(1, 3, figsize=(16, 4))

sns.histplot(df["width"], bins=25, color="steelblue", ax=axes[0])
axes[0].set_title("Image Width Distribution")
axes[0].set_xlabel("Width (Pixels)")

sns.histplot(df["height"], bins=25, color="coral", ax=axes[1])
axes[1].set_title("Image Height Distribution")
axes[1].set_xlabel("Height (Pixels)")

sns.histplot(df["aspect_ratio"], bins=25, color="seagreen", ax=axes[2])
axes[2].set_title("Image Aspect Ratio Distribution")
axes[2].set_xlabel("Width / Height")

plt.tight_layout()
plt.show()
```

The image dimensions vary across the dataset. Most images have heights close to 600 pixels, while widths are more spread out, with a large concentration near 750 pixels. The aspect ratios show that the dataset contains both portrait and landscape paintings. This suggests that resizing will be necessary before training the CNN to ensure consistent image dimensions.
